Run the pipeline first:

```
from pathlib import Path
import xarray as xr

ROOT = Path('..').resolve()
interim = ROOT/'data'/'interim'
zarr_store = ROOT/'data'/'zarr'/'metref_sample_1month.zarr'

print('Interim dir:', interim)
print('Zarr store:', zarr_store)
```

```
files = sorted(interim.rglob('*.nc'))
print('Interim files:', len(files))
files[:3]
```

```
assert files, 'No interim NetCDF files found. Did preprocessing run?'

ds = xr.open_dataset(files[0])
ds
```

```
assert zarr_store.exists(), 'Zarr store not found. Did cube build run?'

ds_z = xr.open_zarr(zarr_store, consolidated=True)
ds_z
```

```
# Small computation to validate dask/chunks
var = list(ds_z.data_vars)[0]
val = float(ds_z[var].isel(time=0).mean().compute())
print('Mean of', var, 'at time=0:', val)
```
